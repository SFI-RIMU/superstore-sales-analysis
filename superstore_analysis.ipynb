```python
import pandas as pd
from pathlib import Path

file_path = Path("D:\Superstore_Sales_Analysis\data\superstore_sales.csv")

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicated rows:")
print(df.duplicated().sum())





```

```python
#checking rows where order date is missing
print("Rows with missing Order Date:")
display(df[df["Order Date"].isna()].head(10))

#checking rows where Sales is missing
print("\nRows with missing Sales:")
display(df[df["Sales"].isna()].head(10))

#checking Duplicate rows
print("\nSample duplicate rows:")
display(df[df.duplicated(keep=False)].head(10))

#Counting rows with missing Order Date
print("\nMissing Order Date count:",df["Order Date"].isna().sum())

#Counting rows with missing Sales
print("Missing Sales count:",df["Sales"].isna().sum())


```

```python
#finding rows with only actual Order Dates
df_clean = df[df["Order Date"].notna()].copy()

#converting date columns (text or object) to datetime
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"], errors="coerce"
)
df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"], errors="coerce"
)   

#removing duplicate rows
df_clean = df_clean.drop_duplicates()

#cleaned dataset checking
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate rows after cleaning:", df_clean.duplicated().sum())

```

```python
#finding or extracting year and month from Order Date
df_clean["Year"] = df_clean["Order Date"].dt.year
df_clean["Month"] = df_clean["Order Date"].dt.month

#creating a yr-month column for analysing monthly sales
df_clean["Year-Month"] = df_clean["Order Date"].dt.to_period("M")

#result
display(
    df_clean[
        ["Order Date", "Year", "Month", "Year-Month", "Sales", "Profit"]
    ].head(10)
)
```

## Core Business Performance Indicators
This section analyzes the overall sales, profit, quantity sold, and average discount in the cleaned dataset to understand the business's general performance.

```python
#calculate overall core business performance
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_quantity = df_clean["Quantity"].sum()
average_discount = df_clean["Discount"].mean()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity:{total_quantity:,}")
print(f"Average Discount:{average_discount:,.2%}")
```

## Sales and Profitability by Category
This section compares total sales and profit accross product categories to identify the strongest and weakest performing categories.

```python
#calculating sales and profit by category
category_analysis = df_clean.groupby("Category").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum")
).sort_values(by="total_sales", ascending=False)

display(category_analysis.round(2))
```

## category performance visualization
Bar charts are used to compare total sales and total profit across the three product categories.

```python
import matplotlib.pyplot as plt
import seaborn as sns

#preparing category data for plotting
category_plot = category_analysis.reset_index()

#chart
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns .barplot(
    data=category_plot,
    x="Category",
    y="total_sales",
    hue="Category",
    legend=False,
    ax=axes[0]
)

axes[0].set_title("Total Sales by Category")
axes[0].set_xlabel("Category")
axes[0].set_ylabel("Total Sales($)")
axes[0].tick_params(axis="x", rotation=15)

sns .barplot(
    data=category_plot,
    x="Category",
    y="total_profit",
    hue="Category",
    legend=False,
    ax=axes[1]
)

axes[1].set_title("Total Profit by Category")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("Total Profit($)")
axes[1].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()

```

### Key Findings

1. Technology generated the highest total sales and profit among the three product categories.
2. Office Supplies generated the second-highest profit despite having the lowest total sales.
3. Furniture had substantial sales but significantly lower profit than Technology and Office Supplies, suggesting that its profitability needs further investigation.

## Monthly Seasonality

This section analyzes monthly sales over time to identify sales patterns and potential seasonal trends.

```python
#total sales for each month
monthly_sales = df_clean.groupby("Year-Month")["Sales"].sum().reset_index()

#converting yr-month to string for plotting
monthly_sales["Year-Month"] = monthly_sales["Year-Month"].astype(str)

#plotting monthly sales
plt.Figure(figsize=(15, 6))
sns.lineplot(
    data=monthly_sales,
    x="Year-Month",
    y="Sales",
    marker="o",
    linewidth=2,
    markersize=4
    )

plt.title("Monthly Sales Trend (2015-2018)", fontsize=16, pad=15,)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)

#SHOW every 3rd month to avoid crowded labels
plt.xticks(
    ticks=range(0, len(monthly_sales), 3),
    labels=monthly_sales["Year-Month"].iloc[::3],
    rotation=45,
    ha="right"
)

plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show
```

### Discount Sensitivity
This section examines the relationship between discount rates and profit to understand how discounts may affect business profitability.

```python
#plot discount vs profit
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha= 0.5
)

plt.title("Discount vs Profit")
plt.xlabel("Discount Rate")
plt.ylabel("Profit ($)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#calculating sales and profit at each discount level
discount_analysis = df_clean.groupby("Discount").agg(
    total_sales = ("Sales", "sum"),
    total_profit = ("Profit", "sum"),
    average_profit = ("Profit", "mean"),
    Number_of_Orders = ("Order ID", "count"),
).round(2)

display(discount_analysis)
```

### Region-wise Sales and Profit

This section compares sales and profit across regions to identify regional performance differences and potential opportunities for business improvement.

```python
#calculating sales and profit by region
region_analysis = df_clean.groupby("Region").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit","sum"),
    total_quantity=("Quantity", "sum")
).sort_values(by="total_sales", ascending=False)

display(region_analysis.round(2))
```

### Customer Segment Analysis

This section compares sales and profit across customer segments to identify the most valuable customer groups.

```python
#calculating sales and profit by customer segment
segment_analysis=df_clean.groupby("Segment").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum")
).sort_values(by="total_sales", ascending=False)

display(segment_analysis.round(2))

```

## Executive Summary and Business Recommendations

### Key Business Findings

1. **Overall Performance:** The business generated total sales of $2,297,200.86 and total profit of $286,397.02 across 9,994 sales records.

2. **Category Performance:** Technology achieved the highest sales ($836,154.03) and profit ($145,454.95). Furniture generated $741,999.80 in sales but only $18,451.27 in profit, indicating a need to investigate its profitability.

3. **Monthly Sales Trends:** Monthly sales fluctuated over time, with a visible peak in November 2018. Further year-over-year comparisons would help confirm recurring seasonal patterns.

4. **Discount and Profitability:** Several discount levels of 30% or more were associated with negative total profit. The business should review high-discount transactions to understand their effect on profitability.

5. **Regional Performance:** The West region generated the highest sales ($725,457.82) and profit ($108,418.45). The Central region generated higher sales than the South but lower total profit.

6. **Customer Segments:** The Consumer segment contributed the highest sales ($1,161,401.34) and profit ($134,119.21), making it the strongest segment by total sales and profit.

### Business Recommendations

- Review Furniture pricing, costs, and discount practices to identify opportunities to improve profit margins.
- Evaluate high-discount transactions and introduce discount limits or approval rules where appropriate.
- Investigate the Central region's lower profit relative to sales and compare its product mix and discount patterns with other regions.
- Maintain customer engagement in the Consumer segment while exploring opportunities to grow Corporate and Home Office sales.
- Use historical monthly sales patterns to support inventory planning and future promotional campaigns.
